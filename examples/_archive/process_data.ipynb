```python
from pathlib import Path
from surfalize import Surface
from surfalize.addons_bv import operation_batch
import numpy as np

file_path = Path(r'E:\LocalData\Programming\Python\surfalize\data\Si_55fs_1.80Jcm2_005Pulses_1.plux')
```

```python
from surfalize.addons_bv.parameters import find_str_x_center

surf = Surface.load(file_path)
find_str_x_center(surf, str_period_um=5.319, filter_radius=20, orders=3, plot_results=False)
```

```python
surf = Surface.load(file_path)
surf.level(inplace=True)
surf.fill_nonmeasured(inplace=True)
surf.show()
operation_batch.fft_filter_periodic(surf, type='pass', str_period_um=5.319, filter_radius=20, orders=3, plot_fft=False)
surf.show()
```

```python
from surfalize.addons_bv.parameters import calculate_morph_depth

_, _, _, _, s_eroded, s_dilated, s_diff = calculate_morph_depth(surf,5.5, crop_edges=False)
import matplotlib.pyplot as plt
_ ,ax = s_eroded.plot_2d()
ax.set_title('Eroded Surface')
_, ax = s_dilated.plot_2d()
ax.set_title('Dilated Surface')
_, ax = s_diff.plot_2d()
ax.set_title('Difference Surface')
```

```python
from surfalize.addons_bv.help_fncs import fit_gaussian_peak

# Use the function
peak_position, stddev, popt = fit_gaussian_peak(s_diff)
```

```python
operation_batch.crop_at_x_position(surf, center_x=peak_position, crop_width=600, debug_info=True)
surf.show()
```
